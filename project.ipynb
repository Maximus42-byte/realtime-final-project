```python
# -*- coding: utf-8 -*-

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from dataclasses import dataclass
from typing import List, Dict, Tuple, Optional
from enum import Enum
import random
import time
import pandas as pd
import copy
from scipy.optimize import linprog
import logging
import warnings

warnings.filterwarnings('ignore')
random.seed(42)
np.random.seed(42)

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)



class SchedulingPolicy(Enum):
    """Scheduling policies for task execution"""
    FIFO = "FIFO"
    HEFT = "HEFT"   # Heterogeneous Earliest Finish Time (rank-based)
    LDF  = "LDF"    # Latest Deadline First


class AllocationPolicy(Enum):
    """Core allocation policies for partitioned scheduling"""
    FFD = "First-Fit Decreasing"
    BFD = "Best-Fit Decreasing"
    WFD = "Worst-Fit Decreasing"
    RR  = "Round-Robin"


class SchedulingType(Enum):
    """Types of scheduling approaches"""
    GLOBAL = "Global"
    PARTITIONED = "Partitioned"


class PreemptionMode(Enum):
    PREEMPTIVE = "Preemptive"
    NON_PREEMPTIVE = "Non-Preemptive"


# ----------------------- Data models -----------------------

@dataclass
class Bundle:
    """Represents a task bundle with variable core requirements"""
    bundle_id: int
    wcet: float                  # Worst Case Execution Time (time units)
    core_requirement: int        # Cores needed concurrently

    def __post_init__(self):
        if self.core_requirement <= 0:
            raise ValueError("Bundle.core_requirement must be positive")
        if self.wcet <= 0:
            raise ValueError("Bundle.wcet must be positive")


@dataclass
class Task:
    """Represents a bundled task in the real-time system"""
    task_id: int
    period: float
    deadline: float
    bundles: List[Bundle]
    priority: int = 0
    arrival_time: float = 0.0

    def __post_init__(self):
        if not self.bundles:
            raise ValueError("Task must have at least one bundle")
        self.total_wcet = sum(b.wcet for b in self.bundles)
        self.max_cores = max(b.core_requirement for b in self.bundles)
        self.utilization = self.total_wcet / self.period

    def get_bundle_count(self) -> int:
        return len(self.bundles)


# ------------------- Task set generation -------------------

class TaskSetGenerator:
    """Generates task sets with specified characteristics"""

    @staticmethod
    def generate_task_set(n_tasks: int,
                          target_utilization: float,
                          max_cores: int,
                          max_bundles: int = 5) -> List[Task]:
        """
        Generate a task set. Keeps per-task complexity sane across different n.
        """
        tasks: List[Task] = []

        base_task_util = 0.05  # 5% baseline per task (before scaling)
        utils = []
        for _ in range(n_tasks):
            u = base_task_util * random.uniform(0.8, 1.2)   # ±20%
            utils.append(min(u, 0.15))                     # cap at 15%

        current_total = sum(utils)
        if current_total > 0:
            # scale toward target, but avoid extreme downscaling
            scale = min(2.0, max(0.05, target_utilization / current_total))
            utils = [max(1e-5, u * scale) for u in utils]  # keep strictly > 0

        for i in range(n_tasks):
            period_base = random.choice([100, 200, 500, 1000])
            period = period_base * random.uniform(0.8, 1.2)

            n_bundles = random.randint(1, max_bundles)

            bundles: List[Bundle] = []
            remaining = utils[i]
            for j in range(n_bundles):
                if j == n_bundles - 1:
                    bundle_util = max(1e-6, remaining)      # ensure > 0
                else:
                    take = remaining * random.uniform(0.2, 0.6)
                    bundle_util = max(1e-6, take)
                    remaining -= bundle_util

                wcet = bundle_util * period
                core_req = random.randint(1, min(max_cores, 4))
                bundles.append(Bundle(j, wcet, core_req))

            # Deadlines: some tighter/looser than period
            deadline = period * random.uniform(0.8, 1.3)
            tasks.append(Task(i, period, deadline, bundles))

        total_util = sum(t.utilization for t in tasks)
        logger.info(f"Generated {n_tasks} tasks (target util ~{target_utilization:.2f}, actual {total_util:.3f})")
        return tasks


# --------------- Response-time analysis (heuristic) ---------------

class ResponseTimeAnalyzer:
    """Implements (heuristic) response time analysis methods"""

    def __init__(self, num_cores: int):
        self.num_cores = num_cores

    # --- policy “effects” (lightweight modeling of ordering efficiency) ---

    def _apply_policy_effects(self, tasks: List[Task],
                              policy: SchedulingPolicy,
                              response_times: Dict[int, float]) -> Dict[int, float]:
        if policy == SchedulingPolicy.HEFT:
            # give a small improvement to tasks ranked high by HEFT
            for t in tasks:
                if hasattr(t, 'bottom_level'):
                    response_times[t.task_id] *= 0.90  # ~10% better

        elif policy == SchedulingPolicy.LDF:
            # LDF tends to harm tight-deadline tasks
            for t in tasks:
                if t.deadline <= t.period:
                    response_times[t.task_id] *= 1.08
                else:
                    response_times[t.task_id] *= 1.00

        # FIFO = baseline (no change)
        return response_times

    # --- “closed-form” (heuristic) ---

    def closed_form_analysis(self,
                             tasks: List[Task],
                             scheduling_type: SchedulingType,
                             preemption: PreemptionMode,
                             policy: Optional[SchedulingPolicy] = None) -> Dict[int, float]:
        rts: Dict[int, float] = {}

        # order by assigned priority (lower number means higher priority)
        ordered = sorted(tasks, key=lambda t: t.priority)
        for t in ordered:
            if scheduling_type == SchedulingType.GLOBAL:
                rt = self._global_response_time(t, ordered, preemption)
            else:
                rt = self._partitioned_response_time(t, ordered, preemption)
            rts[t.task_id] = rt

        if policy is not None:
            rts = self._apply_policy_effects(tasks, policy, rts)
        return rts

    def _compute_blocking_global(self, task: Task, all_tasks: List[Task]) -> float:
        """Non-preemptive blocking term (global): max bundle of any lower-priority task."""
        lower = [x for x in all_tasks if x.priority > task.priority]
        if not lower:
            return 0.0
        return max((max(b.wcet for b in t.bundles) for t in lower), default=0.0)

    def _global_response_time(self, task: Task, all_tasks: List[Task], preemption: PreemptionMode) -> float:
        base = sum(b.wcet for b in task.bundles)

        # higher priority tasks only
        hp = [x for x in all_tasks if x.priority < task.priority and x.task_id != task.task_id]

        # Interference: each HP task may recur within task’s deadline
        interference = 0.0
        total_hp_cores = sum(x.max_cores for x in hp)
        contention = 1.0 + min(1.0, total_hp_cores / max(1, self.num_cores))

        for x in hp:
            k = max(1, int(task.deadline / x.period))      # # of releases
            interference += k * x.total_wcet * contention

        # mild system overhead grows with load
        util = sum(x.utilization for x in all_tasks)
        queueing = base * max(0.0, (util - 0.7)) * 1.5
        overhead = 0.01 * len(all_tasks)

        blocking = 0.0
        if preemption == PreemptionMode.NON_PREEMPTIVE:
            blocking = self._compute_blocking_global(task, all_tasks)

        return base + interference + queueing + overhead + blocking

    def _compute_blocking_part(self, task: Task, all_tasks: List[Task], alloc: Dict[int, List[int]]) -> float:
        """Non-preemptive blocking (partitioned): max bundle of any lower-priority task sharing cores."""
        my_cores = alloc.get(task.task_id, [])
        if not my_cores:
            return 0.0
        lower = [x for x in all_tasks if x.priority > task.priority]
        worst = 0.0
        for x in lower:
            other = set(alloc.get(x.task_id, []))
            if other & set(my_cores):
                worst = max(worst, max(b.wcet for b in x.bundles))
        return worst

    def _partitioned_response_time(self, task: Task, all_tasks: List[Task], preemption: PreemptionMode) -> float:
        # optimistic: when bundle fits cores, it runs at its wcet
        base = 0.0
        for b in task.bundles:
            if b.core_requirement <= self.num_cores:
                base += b.wcet
            else:
                base += b.wcet * (b.core_requirement / self.num_cores)

        # local interference (probabilistic co-location proxy)
        local_int = 0.0
        for x in all_tasks:
            if x.priority < task.priority and x.task_id != task.task_id:
                util_pressure = task.utilization + x.utilization
                core_pressure = (task.max_cores + x.max_cores) / max(1, self.num_cores)
                coloc_p = min(0.8, util_pressure * core_pressure)
                if random.random() < coloc_p:
                    local_int += x.total_wcet * min(0.7, x.utilization)

        sync_overhead = 0.02 * len(task.bundles) * task.max_cores
        blocking = 0.0  # baseline (no allocation-aware info here)
        if preemption == PreemptionMode.NON_PREEMPTIVE:
            # worst-case lower-priority single bundle (coarse upper bound)
            lower = [x for x in all_tasks if x.priority > task.priority]
            if lower:
                blocking = max(max(b.wcet for b in x.bundles) for x in lower)

        return min(base + local_int + sync_overhead + blocking, task.deadline * 2)

    # --- “MILP” (actually a tiny LP lower bound here, documented) ---

    def milp_analysis(self, tasks: List[Task]) -> Dict[int, float]:
        """
        Optimistic *lower-bound* LP. We keep the name to match the spec,
        but document that it's an optimistic bound, not a full MILP schedule.
        """
        rts: Dict[int, float] = {}
        for t in tasks:
            try:
                m = len(t.bundles)
                if m == 1:
                    rts[t.task_id] = t.bundles[0].wcet * 1.05
                    continue

                c = [1.0] * m
                A_ub = []
                b_ub = []

                for i, b in enumerate(t.bundles):
                    row = [0.0] * m
                    row[i] = 1.0
                    A_ub.append(row)
                    b_ub.append(b.wcet * 1.1)  # individual upper bound

                if m > 1:
                    A_ub.append([b.core_requirement for b in t.bundles])
                    b_ub.append(self.num_cores * 1.2)  # mild oversub allowance

                bounds = [(b.wcet, b.wcet * 1.3) for b in t.bundles]

                res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds,
                              method='highs', options={'disp': False})

                if res.success and res.x is not None:
                    base = sum(res.x)
                    rts[t.task_id] = base + 0.05 * t.total_wcet
                else:
                    rts[t.task_id] = t.total_wcet * 1.2
            except Exception:
                rts[t.task_id] = t.total_wcet * 1.2
        return rts

    # --- allocation-aware path (Partitioned) ---

    def analyze_with_allocation(self, tasks: List[Task],
                                scheduling_type: SchedulingType,
                                preemption: PreemptionMode,
                                policy: SchedulingPolicy,
                                core_allocation: Dict[int, List[int]]) -> Dict[int, float]:
        if scheduling_type != SchedulingType.PARTITIONED or not core_allocation:
            return self.closed_form_analysis(tasks, scheduling_type, preemption, policy)

        rts: Dict[int, float] = {}
        for t in tasks:
            alloc = core_allocation.get(t.task_id, [])
            rts[t.task_id] = self._partitioned_response_time_with_allocation(t, tasks, preemption, core_allocation, alloc)

        return self._apply_policy_effects(tasks, policy, rts)

    def _partitioned_response_time_with_allocation(self, task: Task,
                                                   all_tasks: List[Task],
                                                   preemption: PreemptionMode,
                                                   all_allocs: Dict[int, List[int]],
                                                   allocated_cores: List[int]) -> float:
        # base with serialization if under-allocated
        base = 0.0
        k = max(1, len(allocated_cores))
        for b in task.bundles:
            if len(allocated_cores) >= b.core_requirement:
                base += b.wcet
            else:
                base += b.wcet * (b.core_requirement / k)

        # interference from tasks that share any of these cores
        inter = 0.0
        for x in all_tasks:
            if x.priority < task.priority and x.task_id != task.task_id:
                other = set(all_allocs.get(x.task_id, []))
                overlap = len(set(allocated_cores) & other)
                if overlap > 0 and len(allocated_cores) > 0:
                    ratio = overlap / len(allocated_cores)
                    releases = max(1, int(task.deadline / x.period))
                    inter += releases * x.total_wcet * ratio

        # non-preemptive blocking only from lower-priority tasks that share cores
        blocking = 0.0
        if preemption == PreemptionMode.NON_PREEMPTIVE:
            blocking = self._compute_blocking_part(task, all_tasks, all_allocs)

        penalty = self._migration_penalty(allocated_cores)
        eff = self._allocation_efficiency(task, allocated_cores)
        return (base + inter + blocking + penalty) * eff

    def _allocation_efficiency(self, task: Task, allocated_cores: List[int]) -> float:
        if not allocated_cores:
            return 1.5
        sorted_ = sorted(allocated_cores)
        contiguous = all(sorted_[i] + 1 == sorted_[i + 1] for i in range(len(sorted_) - 1))
        need = task.max_cores
        have = len(allocated_cores)

        eff = 1.0
        if have < need: eff *= 1.2
        if have > need: eff *= 1.05
        if not contiguous: eff *= 1.1
        return eff

    def _migration_penalty(self, allocated_cores: List[int]) -> float:
        if len(allocated_cores) <= 1:
            return 0.0
        s = sorted(allocated_cores)
        span = s[-1] - s[0] + 1
        return max(0.0, (span - len(allocated_cores)) * 0.02)


# ------------------------ Core allocator ------------------------

class CoreAllocator:
    """Implements various core allocation strategies"""

    def __init__(self, num_cores: int):
        self.num_cores = num_cores

    def _can_place_block(self, core_usage, start, width, add_util) -> bool:
        return all(core_usage[start + i] + add_util <= 1.0 for i in range(width))

    def allocate_tasks(self, tasks: List[Task], policy: AllocationPolicy) -> Dict[int, List[int]]:
        # Sort by demand (greedy-friendly)
        tasks_sorted = sorted(tasks, key=lambda t: (t.max_cores, t.utilization), reverse=True)
        allocation: Dict[int, List[int]] = {}
        core_usage = [0.0] * self.num_cores

        if policy == AllocationPolicy.FFD:
            for t in tasks_sorted:
                placed = False
                for start in range(max(1, self.num_cores - t.max_cores + 1)):
                    if self._can_place_block(core_usage, start, t.max_cores, t.utilization):
                        allocation[t.task_id] = list(range(start, start + t.max_cores))
                        for i in range(t.max_cores):
                            core_usage[start + i] += t.utilization
                        placed = True
                        break
                if not placed:
                    idx = np.argsort(core_usage)[:t.max_cores]
                    if any(core_usage[i] + t.utilization / t.max_cores > 1.0 for i in idx):
                        allocation[t.task_id] = []
                    else:
                        allocation[t.task_id] = list(idx)
                        for i in idx:
                            core_usage[i] += t.utilization / t.max_cores

        elif policy == AllocationPolicy.BFD:
            for t in tasks_sorted:
                best = None
                best_waste = float('inf')
                for start in range(max(1, self.num_cores - t.max_cores + 1)):
                    if self._can_place_block(core_usage, start, t.max_cores, t.utilization):
                        remaining = sum(1.0 - core_usage[start + i] for i in range(t.max_cores))
                        waste = remaining - t.max_cores * t.utilization
                        if 0 <= waste < best_waste:
                            best_waste = waste
                            best = start
                if best is not None:
                    allocation[t.task_id] = list(range(best, best + t.max_cores))
                    for i in range(t.max_cores):
                        core_usage[best + i] += t.utilization
                else:
                    idx = np.argsort(core_usage)[:t.max_cores]
                    if any(core_usage[i] + t.utilization / t.max_cores > 1.0 for i in idx):
                        allocation[t.task_id] = []
                    else:
                        allocation[t.task_id] = list(idx)
                        for i in idx:
                            core_usage[i] += t.utilization / t.max_cores

        elif policy == AllocationPolicy.WFD:
            for t in tasks_sorted:
                idx = np.argsort(core_usage)[:t.max_cores]
                if any(core_usage[i] + t.utilization / t.max_cores > 1.0 for i in idx):
                    allocation[t.task_id] = []
                else:
                    allocation[t.task_id] = list(idx)
                    for i in idx:
                        core_usage[i] += t.utilization / t.max_cores

        elif policy == AllocationPolicy.RR:
            current = 0
            for t in tasks_sorted:
                picked = []
                tries = 0
                while len(picked) < t.max_cores and tries < 3 * self.num_cores:
                    c = current % self.num_cores
                    if core_usage[c] + t.utilization / max(1, t.max_cores) <= 1.0:
                        picked.append(c)
                        core_usage[c] += t.utilization / max(1, t.max_cores)
                    current += 1
                    tries += 1
                allocation[t.task_id] = picked if len(picked) == t.max_cores else []
        else:
            raise ValueError(f"Unknown allocation policy: {policy}")

        return allocation


# --------------------------- Scheduler ---------------------------

class Scheduler:
    """Runs analyses under chosen policy & scheduling type"""

    def __init__(self, num_cores: int, scheduling_type: SchedulingType):
        self.num_cores = num_cores
        self.scheduling_type = scheduling_type
        self.analyzer = ResponseTimeAnalyzer(num_cores)
        self.allocator = CoreAllocator(num_cores)

    def _assign_priorities(self, tasks: List[Task], policy: SchedulingPolicy):
        if policy == SchedulingPolicy.FIFO:
            for i, t in enumerate(tasks):
                t.priority = i

        elif policy == SchedulingPolicy.LDF:
            for i, t in enumerate(sorted(tasks, key=lambda x: x.deadline, reverse=True)):  # latest first
                t.priority = i

        elif policy == SchedulingPolicy.HEFT:
            # upward rank proxy: compute + comm
            for t in tasks:
                comp = t.total_wcet
                comm = 2.0 * len(t.bundles)
                t.bottom_level = comp + comm
            for i, t in enumerate(sorted(tasks, key=lambda x: x.bottom_level, reverse=True)):
                t.priority = i

    def _estimate_energy(self, tasks: List[Task],
                         policy: SchedulingPolicy) -> float:
        """Simple energy proxy: sum(wcet*cores) adjusted by policy & sched type."""
        energy = 0.0
        for t in tasks:
            for b in t.bundles:
                energy += b.wcet * b.core_requirement

        # scheduling-type factor
        energy *= (0.98 if self.scheduling_type == SchedulingType.PARTITIONED else 1.05)
        # policy factor
        factors = {SchedulingPolicy.FIFO: 1.02,
                   SchedulingPolicy.HEFT: 0.95,
                   SchedulingPolicy.LDF: 1.03}
        energy *= factors.get(policy, 1.0)
        return energy

    def schedule_tasks(self, tasks: List[Task],
                       policy: SchedulingPolicy,
                       preemption: PreemptionMode,
                       allocation_policy: AllocationPolicy = AllocationPolicy.FFD) -> Dict:
        t0 = time.time()

        self._assign_priorities(tasks, policy)

        # Allocation step for Partitioned
        if self.scheduling_type == SchedulingType.PARTITIONED:
            core_alloc = self.allocator.allocate_tasks(tasks, allocation_policy)
            for tid, cores in core_alloc.items():
                for t in tasks:
                    if t.task_id == tid:
                        t.allocated_cores = cores
                        break
        else:
            core_alloc = {}

        # Measure CF and MILP overhead separately
        cf_t0 = time.time()
        if self.scheduling_type == SchedulingType.PARTITIONED:
            cf_rt = self.analyzer.analyze_with_allocation(tasks, self.scheduling_type, preemption, policy, core_alloc)
        else:
            cf_rt = self.analyzer.closed_form_analysis(tasks, self.scheduling_type, preemption, policy)
        cf_t1 = time.time()

        mlp_t0 = time.time()
        milp_rt = self.analyzer.milp_analysis(tasks)
        mlp_t1 = time.time()

        total_util = sum(t.utilization for t in tasks)
        sched_cf = sum(1 for t in tasks if cf_rt.get(t.task_id, 1e30) <= t.deadline) / len(tasks)
        sched_lp = sum(1 for t in tasks if milp_rt.get(t.task_id, 1e30) <= t.deadline) / len(tasks)

        cf_vals = list(cf_rt.values())
        lp_vals = list(milp_rt.values())
        avg_wcet = float(np.mean([t.total_wcet for t in tasks]))
        avg_sched_delay = float(np.mean(cf_vals) - avg_wcet) if cf_vals else 0.0

        energy_est = self._estimate_energy(tasks, policy)

        result = {
            'tasks': tasks,
            'core_allocation': core_alloc,
            'closed_form_response_times': cf_rt,
            'milp_response_times': milp_rt,
            'total_utilization': total_util,
            'avg_total_wcet': avg_wcet,
            'avg_sched_delay': max(0.0, avg_sched_delay),
            'schedulability_ratio_cf': sched_cf,
            'schedulability_ratio_milp': sched_lp,
            'avg_response_time_cf': float(np.mean(cf_vals)) if cf_vals else 0.0,
            'avg_response_time_milp': float(np.mean(lp_vals)) if lp_vals else 0.0,
            'max_response_time_cf': float(np.max(cf_vals)) if cf_vals else 0.0,
            'max_response_time_milp': float(np.max(lp_vals)) if lp_vals else 0.0,
            'energy_estimate': energy_est,
            'analysis_overhead': time.time() - t0,
            'overhead_cf': cf_t1 - cf_t0,
            'overhead_milp': mlp_t1 - mlp_t0,
            'num_cores': self.num_cores,
            'scheduling_type': self.scheduling_type.value,
            'scheduling_policy': policy.value,
            'preemption_mode': preemption.value,
            'allocation_policy': allocation_policy.value if self.scheduling_type == SchedulingType.PARTITIONED else None,
        }
        return result


# ----------------------- Experiment runner -----------------------

class ExperimentRunner:
    def __init__(self):
        self.results: List[Dict] = []
        self.core_configs = [4, 8, 16, 32]
        self.task_counts = [10, 50, 100, 200, 400, 600]
        self.utilization_levels = [0.25, 0.5, 0.75]
        self.scheduling_policies = [SchedulingPolicy.FIFO, SchedulingPolicy.HEFT, SchedulingPolicy.LDF]
        self.allocation_policies = [AllocationPolicy.FFD, AllocationPolicy.BFD,
                                    AllocationPolicy.WFD, AllocationPolicy.RR]
        self.preemption_modes = [PreemptionMode.PREEMPTIVE, PreemptionMode.NON_PREEMPTIVE]

    def _stamp(self, result: Dict, n_tasks: int, util: float):
        result['n_tasks'] = n_tasks
        result['target_utilization'] = util

    def run_comprehensive(self) -> List[Dict]:
        base_total = len(self.core_configs) * len(self.task_counts) * len(self.utilization_levels)
        per_base_runs = len(self.scheduling_policies) * len(self.preemption_modes) * (1 + len(self.allocation_policies))
        total_runs = base_total * per_base_runs
        print(f"=== Starting comprehensive experiments ===")
        print(f"Base configs: {base_total}, Runs per base: {per_base_runs}, Total runs: {total_runs}")
        cnt = 0

        for ncores in self.core_configs:
            for ntasks in self.task_counts:
                for util in self.utilization_levels:
                    base_tasks = TaskSetGenerator.generate_task_set(ntasks, util, ncores)
                    for pol in self.scheduling_policies:
                        for prem in self.preemption_modes:
                            # Global
                            cnt += 1
                            print(f"[{cnt}/{total_runs}] Global | cores={ncores}, tasks={ntasks}, util={util}, pol={pol.value}, prem={prem.value}")
                            sch = Scheduler(ncores, SchedulingType.GLOBAL)
                            res = sch.schedule_tasks(copy.deepcopy(base_tasks), pol, prem)
                            self._stamp(res, ntasks, util)
                            self.results.append(res)

                            # Partitioned (each allocation policy)
                            for ap in self.allocation_policies:
                                cnt += 1
                                print(f"[{cnt}/{total_runs}] Part   | cores={ncores}, tasks={ntasks}, util={util}, pol={pol.value}, prem={prem.value}, alloc={ap.value}")
                                schp = Scheduler(ncores, SchedulingType.PARTITIONED)
                                resp = schp.schedule_tasks(copy.deepcopy(base_tasks), pol, prem, ap)
                                self._stamp(resp, ntasks, util)
                                self.results.append(resp)

        print(f"=== Completed experiments: {len(self.results)} runs ===")
        return self.results


# --------------------------- Analysis & plots ---------------------------

class ResultAnalyzer:
    def __init__(self, results: List[Dict]):
        self.results = results
        self.df = pd.DataFrame(results)
        plt.style.use('seaborn-v0_8')

    def generate_required_plots(self):
        if self.df.empty:
            print("No data to plot.")
            return

        print("Generating required (6) plots...")
        fig = plt.figure(figsize=(20, 18))

        # (1) Response time variations (Closed-Form vs MILP)
        plt.subplot(3, 2, 1);  self._plot_resp_time_methods()

        # (2) Schedulability across core counts
        plt.subplot(3, 2, 2);  self._plot_sched_vs_cores()

        # (3) Utilization impact on RT & Energy
        plt.subplot(3, 2, 3);  self._plot_util_vs_rt_energy()

        # (4) Scheduling-induced delays by policy & preemption
        plt.subplot(3, 2, 4);  self._plot_sched_delay()

        # (5) Computational overhead comparison (CF vs MILP)
        plt.subplot(3, 2, 5);  self._plot_overhead()

        # (6) Task-count scaling (n=[...])
        plt.subplot(3, 2, 6);  self._plot_task_scaling()

        plt.tight_layout()
        plt.savefig('comprehensive_analysis_fixed.png', dpi=300, bbox_inches='tight')
        plt.show()
        self._summary_and_export()

    # ---- Plot helpers for the 6 expected outputs ----

    def _plot_resp_time_methods(self):
        g = self.df.groupby(['scheduling_policy','scheduling_type','preemption_mode']).agg(
            avg_cf=('avg_response_time_cf','mean'),
            avg_lp=('avg_response_time_milp','mean')
        ).reset_index()
        if g.empty:
            plt.title('Response Time (Methods)'); return
        x = np.arange(len(g)); w = 0.35
        plt.bar(x - w/2, g['avg_cf'], w, label='Closed-Form (Heuristic)')
        plt.bar(x + w/2, g['avg_lp'], w, label='MILP (LP Lower Bound)')
        plt.xticks(x, [f"{r['scheduling_policy']}\n{r['scheduling_type']}\n{r['preemption_mode']}" for _, r in g.iterrows()], rotation=0)
        plt.ylabel('Avg Response Time'); plt.title('(1) Response Time: Methods vs Modes'); plt.legend()

    def _plot_sched_vs_cores(self):
        d = self.df.groupby(['num_cores','preemption_mode']).agg(s=('schedulability_ratio_cf','mean')).reset_index()
        for prem in d['preemption_mode'].unique():
            dd = d[d['preemption_mode']==prem]
            plt.plot(dd['num_cores'], dd['s'], 'o-', label=prem)
        plt.xlabel('Cores'); plt.ylabel('Schedulability'); plt.title('(2) Schedulability vs Cores'); plt.legend(); plt.grid(alpha=.3)

    def _plot_util_vs_rt_energy(self):
        u = self.df.groupby(['target_utilization','preemption_mode']).agg(
            avg_rt=('avg_response_time_cf','mean'),
            avg_energy=('energy_estimate','mean')
        ).reset_index()
        if u.empty:
            plt.title('(3) Utilization Impact'); return
        premodes = u['preemption_mode'].unique()
        ax1 = plt.gca()
        for prem in premodes:
            dd = u[u['preemption_mode']==prem]
            ax1.plot(dd['target_utilization'], dd['avg_rt'], 'o-', label=f'RT - {prem}')
        ax1.set_xlabel('Target Utilization'); ax1.set_ylabel('Avg Response Time'); ax1.grid(alpha=.3)
        ax2 = ax1.twinx()
        for prem in premodes:
            dd = u[u['preemption_mode']==prem]
            ax2.plot(dd['target_utilization'], dd['avg_energy'], 's--', label=f'Energy - {prem}')
        ax1.set_title('(3) Utilization → Response Time & Energy')
        ax1.legend(loc='upper left'); ax2.legend(loc='upper right')
        ax2.set_ylabel('Energy (proxy)')

    def _plot_sched_delay(self):
        d = self.df.groupby(['scheduling_policy','preemption_mode']).agg(delay=('avg_sched_delay','mean')).reset_index()
        if d.empty:
            plt.title('(4) Scheduling-Induced Delay'); return
        x = np.arange(len(d)); w = 0.6
        plt.bar(x, d['delay'], w, alpha=.85)
        plt.xticks(x, [f"{r['scheduling_policy']}\n{r['preemption_mode']}" for _, r in d.iterrows()])
        plt.ylabel('Avg (RT - WCET)'); plt.title('(4) Scheduling-Induced Delay by Policy & Preemption')

    def _plot_overhead(self):
        oh = self.df.groupby(['n_tasks']).agg(cf=('overhead_cf','mean'),
                                              mlp=('overhead_milp','mean')).reset_index()
        if oh.empty:
            plt.title('(5) Computational Overhead'); return
        x = np.arange(len(oh)); w = 0.35
        plt.bar(x - w/2, oh['cf'], w, label='Closed-Form')
        plt.bar(x + w/2, oh['mlp'], w, label='MILP (LP)')
        plt.xticks(x, [str(int(v)) for v in oh['n_tasks']]); plt.xlabel('Number of Tasks')
        plt.ylabel('Analysis Time (s)'); plt.title('(5) Computational Overhead vs Task Count'); plt.legend()

    def _plot_task_scaling(self):
        # fixed config for comparability
        pick = self.df[
            (self.df['num_cores']==8) &
            (self.df['target_utilization']==0.5) &
            (self.df['scheduling_type']=='Global')
        ]
        if pick.empty:
            plt.title('(6) Task-Count Scaling'); plt.text(0.5,0.5,'No data',ha='center',va='center',transform=plt.gca().transAxes); return
        d = pick.groupby(['n_tasks','preemption_mode']).agg(cf=('avg_response_time_cf','mean')).reset_index()
        for prem in d['preemption_mode'].unique():
            dd = d[d['preemption_mode']==prem]
            plt.plot(dd['n_tasks'], dd['cf'], 'o-', label=prem)
        plt.xlabel('Number of Tasks (n)'); plt.ylabel('Avg Response Time')
        plt.title('(6) Task-Count Scaling @ 8 cores, 50% util (Global)'); plt.legend(); plt.grid(alpha=.3)

    def _summary_and_export(self):
        print("\n" + "="*60)
        print("SUMMARY (Required Outputs)")
        print("="*60)
        print(f"Experiments: {len(self.results)}")
        print(f"Cores: {sorted(self.df['num_cores'].unique())}")
        print(f"Tasks: {sorted(self.df['n_tasks'].unique())}")
        print(f"Target Utils: {sorted(self.df['target_utilization'].unique())}")
        print(f"Policies: {sorted(self.df['scheduling_policy'].unique())}")
        print(f"Preemption Modes: {sorted(self.df['preemption_mode'].unique())}")

        # Save CSV with all fields needed for reporting
        cols = ['num_cores','n_tasks','target_utilization','scheduling_type','scheduling_policy',
                'preemption_mode','allocation_policy','avg_total_wcet','avg_sched_delay',
                'avg_response_time_cf','avg_response_time_milp','max_response_time_cf','max_response_time_milp',
                'schedulability_ratio_cf','schedulability_ratio_milp','total_utilization',
                'energy_estimate','analysis_overhead','overhead_cf','overhead_milp']
        self.df[cols].to_csv('detailed_experimental_results_fixed.csv', index=False)
        print("\nSaved files:")
        print("- comprehensive_analysis_fixed.png  (6 charts per project brief)")
        print("- detailed_experimental_results_fixed.csv (raw results)")

        # Key stats
        by_cores = self.df.groupby('num_cores').agg(
            rt=('avg_response_time_cf','mean'),
            sch=('schedulability_ratio_cf','mean')).reset_index()
        print("\nSchedulability by cores:")
        for _, r in by_cores.iterrows():
            print(f"  {int(r['num_cores']):>2} cores -> RT={r['rt']:.3f}, Sched={r['sch']:.3f}")


# -------------------------- Entry points --------------------------

def main_full():
    print("="*60)
    print("REAL-TIME MULTIPROCESSOR BUNDLED TASK SCHEDULING SYSTEM")
    print("FULL SWEEP (Preemptive + Non-Preemptive)")
    print("="*60)
    runner = ExperimentRunner()
    results = runner.run_comprehensive()
    if not results:
        print("No results generated."); return
    analyzer = ResultAnalyzer(results)
    analyzer.generate_required_plots()
    print("\nDone (full).")


if __name__ == "__main__":
    # Full comprehensive experiment (as requested)
    RUN_FULL = True

    if RUN_FULL:
        main_full()
```
